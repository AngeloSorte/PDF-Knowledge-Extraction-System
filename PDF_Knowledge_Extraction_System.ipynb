```python
# PDF Knowledge Extraction System
# Author: Angelo Sorte
# Description: Extract structured data from banking PDF documents

# ------------------------
# Imports
# ------------------------
import PyPDF2
import pandas as pd
import spacy

# ------------------------
# Load NLP model
# ------------------------
nlp = spacy.load("en_core_web_sm")

# ------------------------
# Read PDF
# ------------------------
# Replace 'sample_bank_doc.pdf' with your PDF file path
pdf_file = open('sample_bank_doc.pdf', 'rb')
pdf_reader = PyPDF2.PdfReader(pdf_file)

text = ""
for page in pdf_reader.pages:
    text += page.extract_text()

pdf_file.close()

# ------------------------
# NLP processing
# ------------------------
doc = nlp(text)

# Extract entities
entities = [(ent.text, ent.label_) for ent in doc.ents]

# ------------------------
# Convert to DataFrame
# ------------------------
df = pd.DataFrame(entities, columns=['Entity', 'Type'])
print(df.head())

# ------------------------
# Example: Filter for monetary values and dates
# ------------------------
monetary_values = df[df['Type'] == 'MONEY']
dates = df[df['Type'] == 'DATE']
print("\nMonetary Values:\n", monetary_values)
print("\nDates:\n", dates)
```
